# Tuple

- Ein Tuple speicherte mehrere Werte in einer festen Reihenfolge
- Kann aber nach dem Erstellen nicht mehr geändert werden (im Unterschied zur Liste)
- Tuple wird mit () angelegt

# Tuple anlegen

```python
obst = ("Apfel", "Kirsche", "Birne")

print(obst)
```

# Element zugreifen

```python
print(obst[1])
```

# Tuple ändern

```python
obst[1] = "Banane"
```
