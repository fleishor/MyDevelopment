# List

- Eine Liste speichert mehrere Werte, es kann auch mehrmals der gleiche Wert gespeichert werden.
- Eine Liste ist eine geordnete Sammlung, sprich die Elemente halten ihre Positionen bzw. Reihenfolge bei
- Auf eine Liste kann auch mit einem Index zugegriffen werden
- Wird mit [] angelegt.

# Liste anlegen

```python
obst = ["Apfel", "Kirsche", "Birne"]

print(obst)
```

# Zugriff auf ein Element

```python
print(obst[0])  # Apfel
```

# Elemente hinzufügen und löschen

```python
obst = ["Apfel", "Kirsche", "Birne"]
print(f"Original Liste: {obst}")

obst[1] = "Banane"
print(f"Kirsche durch Banane ersetzt: {obst}")

obst.append("Orange")
print(f"Orange am Ende hinzugefügt: {obst}")

obst.remove("Apfel")
print(f"Apfel gelöscht: {obst}")
```

# Über komplette Liste iterieren

```python
for frucht in obst:
    print(frucht)
```

# Weitere Methoden
| Methode | Wirkung | Beispiel |
|---|---|---|
| `append(x)` | Fügt `x` am Ende hinzu | `liste.append("Birne")` |
| `insert(i, x)` | Fügt `x` an Position `i` ein | `liste.insert(0, "Apfel")` |
| `extend(werte)` | Fügt mehrere Werte an | `liste.extend(["Birne", "Kiwi"])` |
| `remove(x)` | Entfernt das erste Element mit dem Wert `x` | `liste.remove("Apfel")` |
| `pop(i)` | Entfernt und liefert das Element an Position `i` | `liste.pop(0)` |
| `clear()` | Entfernt alle Elemente | `liste.clear()` |
| `index(x)` | Liefert die Position des ersten `x` | `liste.index("Birne")` |
| `count(x)` | Zählt, wie oft `x` vorkommt | `liste.count("Apfel")` |
| `sort()` | Sortiert die Liste | `liste.sort()` |
| `reverse()` | Kehrt die Reihenfolge um | `liste.reverse()` |
| `copy()` | Erstellt eine flache Kopie | `kopie = liste.copy()` |
