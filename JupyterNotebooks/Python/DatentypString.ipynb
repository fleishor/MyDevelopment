# String als iterable

```python
zeichenkette = "hallo"
for zeichen in zeichenkette:
    print(zeichen)
```

```python
vorname = "Horst"
nachname = "Fleischer"
alter = 30

name = vorname + " " + nachname + " " + str(alter)
print(name)

name = f"{vorname} {nachname} {alter}"
print(name)
```

```python
nachname = "Fleischer"

print(nachname[2:4])
print(nachname.find("ei"))
print(nachname.upper())
print(nachname.lower())
print(nachname.strip())
```
