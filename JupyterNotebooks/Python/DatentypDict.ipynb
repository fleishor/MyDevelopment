# Dictionary

- Ein Dictionary (dict) speichert Schlüssel-Wert-Paare. 
- Über den Schlüssel kannst du den zugehörigen Wert abrufen.
- Jeder Schlüssel darf nur einmal vorkommen; ansonsten wird der vorherige Wert überschrieben.
- Wird mit {} angelegt

# Dictionary anlegen

```python
ich = {
    "name": "Horst",
    "alter": 52,
    "wohnort": "Nittenau"
}

print(ich)
```

# Dictionary auslesen

```python
print(ich["name"])

print(ich["familienstand"])
```

# Dictionary ohne Fehlermeldung auslesen

```python
print(ich.get("familienstand", "unbekannt"))
```

# Eintrag in Dictionary ändern bzw. hinzufügen

```python
ich["alter"] = 53
ich["familienstand"] = "verheiratet"

print(ich)
```

# Über komplettes Dictionary iterieren

```python
for schluessel, wert in ich.items():
    print(schluessel, wert)
```
