# For-in-Schleife

- Eine for-in-Schleife durchläuft die Elemente einer iterierbaren Sammlung nacheinander
- Iterierbare Sammlung sind unter Python:
  - Listen (list): ["rot", "blau"]
  - Tupel (tuple): (10, 20, 30)
  - Zeichenketten (str): "Hallo" — die Schleife geht Buchstabe für Buchstabe durch
  - Mengen (set): {"rot", "blau"} — die Reihenfolge ist nicht festgelegt
  - Dictionary (dict) — standardmäßig wird über die Schlüssel iteriert
  - Ranges (range): range(5) — liefert Zahlen von 0 bis `4
  - Dateien — etwa Zeile für Zeile
  - Generatoren — liefern Werte nach und nach, statt alle auf einmal zu speichern

# Iterate Liste

```python
fruechte = ["Apfel", "Kirsche", "Birne"]

for frucht in fruechte:
    print(frucht)
```

# Iterate Dictionary

```python
for schluessel in {"name": "Mia", "alter": 12}:
    print(schluessel)
```

# Iterate Range

```python
for zahl in range(1, 4):
    print(zahl)
```
