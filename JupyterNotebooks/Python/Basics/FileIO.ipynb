# Datei komplett einlesen

datei.read() liest den Inhalt der Datei in einen String

```python
with open("eingabe.txt", "r", encoding="utf-8") as datei:
    inhalt = datei.read()

print(inhalt)
```

# Datei zeilenweise einlesen

Datei zählt zu den "iterierbaren Objekten", somit kann es in einer for-Schleife verwendet werden.

```python
with open("eingabe.txt", "r", encoding="utf-8") as datei:
    for zeile in datei:
        print(f"{zeile}", end="")
```

# Datei zeilenweise einlesen und durchiterieren

enumerate() liefert neben der Zeile auch eine laufende Nummer zurück.

```python
with open("eingabe.txt", "r", encoding="utf-8") as datei:
    for nummer, zeile in enumerate(datei, start=1):
        print(f"{nummer:3}: {zeile}", end="")
```

# String in Datei schreiben

```python
with open("ausgabe.txt", "a", encoding="utf-8") as datei:
    datei.write(inhalt)
```
