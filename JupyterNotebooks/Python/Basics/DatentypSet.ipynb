# Set

- Ein Set (Menge) speichert mehrere Werte, jeder Wert kann dabei nur einmal vorkommen
- Ein Set ist eine ungeordnete Sammlung, sprich die Elemente halten ihre Positionen bzw. Reihenfolge **nicht** bei
- Auf ein Set kann **nicht** auch mit einem Index zugegriffen werden
- Wird mit {} angelegt; analog zum dictionary

# Set anlegen
- Ein Set enthält keine doppelten Einträge

```python
farben = {"rot", "grün", "blau"}
print(f"Farben: {farben}")

farben = {"rot", "grün", "blau", "rot"}
print(f"Farben (rot ist nur einmal enthalten): {farben}")
```

# Elemente löschen und hinzufügen
- Elemente werden irgendwo hinzugefügt (ungeordnete Sammlung)

```python
farben = {"rot", "grün", "blau"}
print(f"Farben: {farben}")

farben.add("orange")
print(f"Orange hinzugefügt: {farben}")

farben.remove("blau")
print(f"blau entfernt: {farben}")
```
